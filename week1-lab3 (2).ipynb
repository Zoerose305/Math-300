5) write a loop that keeps a list of the sum of the first n integers as n ranges from 1 to 30. 

```python
sums = [] 
running_total = 0 

for n in range(1,31): 
     running_total += n
     sums.append(running_total) 
print(sums)
```

6) write a for loop that counts the number of even integers in a list.

```python
numbers = [1,13,4,15,2,7,8]
even_count = 0

for num in numbers: 
    if num % 2 == 0:
        even_count += 1
print(even_count)
```

7) if we list all natural numbers below 10 that are multiples of 3 or 5 we get 3,5,6,9. The sums of these numbers is 23. Find the sum of all multiples of 3 or 5 under 1000. 

```python
total = 0
for num in range (1000):
    if num % 3 == 0 or num % 5 == 0: 
        total += num 
print(total)
```
