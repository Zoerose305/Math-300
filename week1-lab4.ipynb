4) Define a function that takes a maximum number as the input and outputs the sum  of multiples of 3 or 5 below this number.  

```python
def multisum(maximum):
    current = 0
    for i in range(maximum):
        if (i%3 == 0) | (i%5 == 0):
            current += i
        return current
    multisum(10)
    
```

6) define a function. Palindrome(string) that takes in a string and outputs whether or not the string is a palindrome. 

```python
def palindrome(string):
    n = len(string)
    count = 0
    for i in range(n//2):
        if string[i] == string[n-1-i]:
            count += 1
        if count == n//2:
            print('palindrome')
        else:
            print('not palindrome') 
```
