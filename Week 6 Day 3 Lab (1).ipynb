## Problem 1

Using the model from class, replace MobileNetV2 with ResNet50.

Evaluate:

- Training time
- Accuracy
- Model size


```python
dataset_path = tf.keras.utils.get_file(
    "flower photos",
    "https://storage.googleapis.com/download.tensorflow.org/example_images/flower_photos.tgz",
    untar=True
)
```

```python
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
train_dataset = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)
validation_dataset = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)
```

```python
normalization_layer = tf.keras.layers.Rescaling(1.0/ 225)
```

```python
train_dataset = train_dataset.prefetch(tf.data.AUTOTUNE)
```

```python
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(224,224,3),
    include_top=False,
    weights="imagenet"
)
```

```python
base_model.trainable = False
```

```python
model = tf.keras.Sequential([
    base_model,
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(5, activation="softmax")
])
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```


## Problem 2

Create a transfer learning model using the TensorFlow Flowers dataset and classify five flower species.


```python
history = model.fit(
    train_dataset,
    validation_data=validation_dataset,
    epochs=5
)
```

```python
loss, accuracy = model.evaluate(validation_dataset)
```
