### Problem 2
Create your own dataset and train the model for:

- 50 epochs
- 100 epochs
- 300 epochs

Compare the results.


````python
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
], dtype=float)
y =np.array([0,0,0,1])
epochs_list = [50, 100, 300]
for e in epochs_list:
    print(f"\nTraining for {e} epochs")
    model = Sequential([
        Dense(8, input_shape=(2,), activation="relu"),
    Dense(1, activation="sigmoid")
    ])
    model.compile(optimizer="adam",
                loss="binary_crossentropy",
                metrics=["accuracy"])
    model.fit(X,y, epochs=e, verbose=0)
    loss, acc = model.evaluate(X, y, verbose=0)
    print("Loss:", round(loss,4))
    print("Accuracy:", round(acc,4))
    predictions = model.predict(X, verbose=0)
    print("Predictions:")
    print(np.round(predictions,3))
````

### Problem 3

Replace the optimizer:

```python
optimizer="sgd"
```

Compare it with:

```python
optimizer="adam"
```


````python
#optimizer='sgd'
model_sgd = Sequential([
    Dense(8, input_shape=(2,), activation="relu"),
    Dense(1, activation="sigmoid")
])
model_sgd.compile(
    optimizer="sgd",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
model_sgd.fit(X, y, epochs=100, verbose=0)
loss_sgd, acc_sgd = model_sgd.evaluate(X, y, verbose=0)
print("SGD Accuracy:", acc_sgd)
````

````python
#optimizer="adam"
model_adam = Sequential([
    Dense(8, input_shape=(2,), activation="relu"),
    Dense(1, activation="sigmoid")
])
model_adam.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
model_adam.fit(X, y, epochs=100, verbose=0)
loss_adam, acc_adam = model_adam.evaluate(X, y, verbose=0)
print("Adam Accuracy:", acc_adam)
````
