### Problem 1
Write your own function performs vectors addition on lists instead of arrays.

```python
def vector_add(list1, list2):
    if len(list1) != len(list2):
        raise ValueError("Vectors must have same length")
    result = []
    for i in range(len(list1)):
        result.append(list1[i] + list2[i])
    return result
v1 = [1, 2, 3]
v2 = [4, 5, 6]
print(vector_add(v1, v2))
```

### Problem 2
Create two $3\times 2$ matrices, A and B. Compute: 
1. A+B
2. AB

```python
import numpy as np
A = np.array([
    [1, 2],
    [3, 4]
])
B = np.array([
    [5, 6],
    [7, 8]
])
print("A + B:")
print(A + B)
print("\nAB:")
print(A @ B)
```

### Problem 3
Make a 2 by 2 invertible matrix, A.
1. Compute its eigenvectors and plot them using matplotlib
2. Compute A times the eigenvectors and plot them on the same plot.
3. What do you notice?

```python
import matplotlib.pyplot as plt
A = np.array([
    [2, 1],
    [1, 2]
])
eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues:")
print(eigenvalues)
print("\nEigenvectors:")
print(eigenvectors)
```

```python
plt.figure(figsize=(6,6))
orgin = np.array([[0,0],[0,0]])
for i in range(2):
    v = eigenvectors[:, i]
    plt.quiver(0, 0, v[0], v[1],
        angles='xy', scale_units='xy',
        scale=1, color='blue',
        label='Eigenvector')
    Av = A @ v
    plt.quiver(0, 0, Av[0], Av[1],
        angles='xy', scale_units='xy',
        scale=1, color='red',
        label='A times eigenvector')
plt.xlim(-5,5)
plt.ylim(-5,5)
plt.grid()
plt.xlabel("x")
plt.ylabel("y")
plt.title("Eigenvectors & A(Eigenvectors)")
plt.show()
```
