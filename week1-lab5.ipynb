1) The sum of the squares of the first 10 natural numbers is, 385. The square of the sum of the first 10 natural numbers is, 3025. The difference of the sum of the squares of the first 10 natural numbers and the square of the sums is 3025-385 = 2640. Find the difference of the sum of the squares of the first 100 natural numbers and the square of the sum.

```python
sum_nums = sum(range(1,101))
sum_squares = sum(i**2 for i in range(1,101))
difference = sum_nums **2 - sum_squares
print(difference)
```
