### Problem 1
Suppose $f(x)=x^2+6x+10$

1. Plot the function to visualize the minimum
2. Find the minimum using `scipy`

````python
def f(x):
    return x**2 + 6*x + 10
x = np.linspace(-10, 5, 200)
y = f(x)
plt.figure(figsize=(8,5))
plt.plot(x,y)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("f(x)=x²+6x+10")
plt.grid()
plt.show()
result = minimize(f, x0=0)
print("Minimum x:", result.x)
print("Minimum value:", result.fun)
````


## Problem 2

Find the maximum of $50x-x^2$.

````python
def g(x):
    return 50*x - x**2
result = minimize(lambda x: -g(x), x0=0)
print("Maximum x:", result.x)
print("Maximum value:", g(result.x))
````

## Problem 3

Minimize

```python
np.sin(x)+x**2
```

using different starting values.

Do you obtain the same solution? Compute the derivative using `sympy`. When is this derivative zero?

````python
np.sin(x)+x**2
````

````python
def h(x):
    return np.sin(x) + x**2
starting_values = [-5, -2, 0, 2, 5]
for start in starting_values:
    result = minimize(h, x0=start)
    print("Starting value:", start)
    print("Minimum x:", result.x)
    print("Minimum value:", result.fun)
    print()
````

````python
x = sp.Symbol('x')
f = sp.sin(x) + x**2
derivative = sp.diff(f, x)
print("Derivative:")
print(derivative)
````

### Problem 4
Design a function with multiple minima and optimize.

````python
def multi_min(x):
    return np.sin(3*x) + 0.1*x**2
x = np.linspace(-10,10,500)
plt.figure(figsize=(8,5))
plt.plot(x, multi_min(x))
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Function with Multiple Minima")
plt.grid()
plt.show()
````

````python
starts = [-8, -4, 0, 4, 8]
for start in starts:
    result = minimize(multi_min, x0:= start)
    print("start:", start)
    print("Minimum x:", result.x)
    print("minimum value:", result.fun)
    print()
````
